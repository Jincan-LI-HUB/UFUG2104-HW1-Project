**Section 2:** Working with two joint random variables, i.e., petal length and width.

Below, we visualize the joint behavior of two random variables by plotting a 2D scatter plot of petal length and width.

We plot the 2D scatter plots using ax.scatter.

```python
# create figure
fig, ax = plt.subplots()

# create scatter plot
ax.scatter(data['Petal length'], data['Petal width'], alpha=0.3)

axlim = [2, 7, 0, 3]
plt.xlabel('Petal length')
plt.ylabel('Petal width')
plt.grid(True)
ax.set_title('Scatter plot of petal length and width')

plt.show()
```

We start by modelling the joint distribution of petal length and width using a 2D Gaussian distribution $f_{X,Y}(x,y)$ with mean vector $m$ and covariance matrix $C$. 

The code below estimates the mean and covariance that best fit the data using the **pandas.DataFrame.mean** and **pandas.DataFrame.cov** functions.

```python
# Compute the mean vector and covariance matrix
m = data[['Petal length','Petal width']].mean()
C = data[['Petal length','Petal width']].cov()
rho = data[['Petal length','Petal width']].corr(method = 'pearson')

print('The mean vector of petal length and width is\n')
print(m)

print('\nThe covariance matrix of petal length and width is\n')
print(C)

print('\nThe correlation coefficient matrix of petal length and width is\n')
print(rho)
```

The code below overlays a contour plot of the 2D Gaussian that best fits the data onto the scatterplot. 

```python
# create a multivariate Gaussian random variable with mean m and covariance matrix C
dist = multivariate_normal(m, C)

# create limits of plot
axlim = [2, 7, 0, 3]

# create set of (x,y) positions to compute the pdf at
x, y = np.meshgrid( np.linspace(axlim[0], axlim[1]), np.linspace(axlim[2], axlim[3]) )
pos = np.dstack((x, y))

# compute the joint Gaussian pdf
f_XY = dist.pdf(pos)

# create the plot
fig, ax = plt.subplots()

# scatter plot
ax.scatter(data['Petal length'], data['Petal width'], alpha=0.3)

# pdf contour plot
ax.contour(x, y, f_XY, colors='b', alpha = 0.7)

# set axes and labels
ax.axis('scaled')
ax.axis(axlim)
plt.xlabel('Petal length')
plt.ylabel('Petal width')
plt.grid(True)
ax.set_title('The joint PDF')

plt.show()
```

Now, we generate the scatter plot of petal length and width for each species separately. We separate data into different groups using the **pandas.DataFrame.groupby** command.

**Section 3:** Pattern classification with petal length and width

A simple classifier classifies flowers as virginica if the petal length $L$ is greater threshold $T$, and as versicolor otherwise. 

The probability that this classifier makes an error is given by

$
\mathrm{P}[\mathrm{versicolor\ and\ } L>T] + \mathrm{P}[\mathrm{virginica \ and \ } L \leq T] 
= \mathrm{P} [ L > T | \mathrm{versicolor} ] \mathrm{P} [ \mathrm{versicolor} ] 
+ \mathrm{P} [ L \leq T | \mathrm{virginica} ] \mathrm{P} [ \mathrm{virginica} ]
$

Below, we plot the probability that this classifier makes an error for $t$ from 2 to 7 under the following assumtions:

1. The conditional probability density functions of petal length given species, $f_{L | C } (l | c ) $, 
are given by Gaussian distributions

2. The prior probabilities of the two species are equal, i.e. 
$\mathrm{P} [ \mathrm{versicolor} ] = \mathrm{P} [ \mathrm{virginica} ] = 0.5$

We calculate the probabilities that $L > T$ and $L \leq T$ using the survival functions and the cumulative distribution functions.


The plot shows that there is an optimal threshold at $T = 1.6$. The corresponding probability of making an error is 0.07.

```python
groups = data.groupby('Class')
m_pw_by_species = groups['Petal length'].mean()
s_pw_by_species = groups['Petal length'].std()

t = np.linspace(2, 7, 100)

p_pw_e_versicolor = scipy.stats.norm.sf(t, loc = m_pw_by_species['Iris-versicolor'], scale = s_pw_by_species['Iris-versicolor'])
p_pw_e_virginica = scipy.stats.norm.cdf(t, loc = m_pw_by_species['Iris-virginica'], scale = s_pw_by_species['Iris-virginica'])
p_pw_e = p_pw_e_versicolor*0.5 + p_pw_e_virginica*0.5

fig = plt.figure()
plt.plot(t, p_pw_e)
plt.title('The probability that this classifier makes an error')
plt.xlabel('petal length')
plt.ylabel('probability')
plt.grid()
plt.show()

print(f'The minmum probability that this classifier makes an error is {min(p_pw_e)}')
print(f'The optimum threshold is {t[np.argmin(p_pw_e)]}')
```
